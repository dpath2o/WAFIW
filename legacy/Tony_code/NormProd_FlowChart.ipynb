```python
#version 1

with schemdraw.Drawing() as d:
    d.config(fontsize=13, unit=.5)

                                                
    # Column 1
    box1 = flow.Terminal().label('Boxcar 1')
    flow.Arrow().length(1.5)
    subtract = flow.Process().label('Subtract boxcar from original image')
    flow.Arrow().length(1.5)
    dob1 = flow.Terminal().label('Difference of Boxcars 1 (DoB 1)')

    # Column 2, placed closer (x=5 instead of 17)
    box2 = flow.Terminal().label('Boxcar 2').at((9, 0))
    flow.Arrow().length(1.5)
    subtract2 = flow.Process().label('Subtract boxcar from original image')
    flow.Arrow().length(1.5)
    dob2 = flow.Terminal().label('Difference of Boxcars 2 (DoB 2)')

    # column 3 (middle) ?
    a = flow.Process().label('Multiply DoBs together').at((4.5, -9))
    flow.Arrow().length(1.5)
    normprod = flow.Terminal().label('NormProd')
    flow.Arrow().length(1.5)
    mean = flow.Process().label('mean')
    flow.Arrow().length(1.5)
    summednormprod = flow.Terminal().label('Meaned_Normprod')

    flow.ArcN(arrow='->').at(dob1.S).to(a.N)
    flow.ArcN(arrow='->').at(dob2.S).to(a.N)


   #collumn 4
    std1 = flow.Terminal().label('Stdev 1').at((20, 0))
    std2 = flow.Terminal().label('Stdev 2').at((27, 0))
    mean2 = flow.Process().label('calculate the mean of the two stdevs').at((23, -3.5))
    flow.Arrow().length(1.5)
    stdmean = flow.Terminal().label('Stdmean')
    flow.Arrow().length(1.5)
    flow.Process().label('square Stdmean')
    flow.Arrow().length(1.5)
    flow.Terminal().label('Variance')
    flow.Arrow().length(1.5)
    boxcar2 = flow.Process().label('Boxcar smoothing')
    flow.Arrow().length(1.5)
    smovar = flow.Terminal().label('Smoothed_Variance')

    flow.ArcN(arrow='->').at(std1.S).to(mean2.N)
    flow.ArcN(arrow='->').at(std2.S).to(mean2.N)

    #final column
    x = flow.Process().label('Meaned_NormProd / Smoothed_Variance').at((14, -23))
    flow.Arrow().length(1.5)
    flow.Terminal().label('NormProd_SmoVar')

    flow.Arc2(arrow='->', k = -0.5).at(summednormprod.S).to(x.W)
    flow.Arc2(arrow='->').at(smovar.S).to(x.E)

    # top columns
    original1 = flow.Terminal().label('Original Image 1').at((4.5,8))
    #flow.Arrow().length(3)
    smooth1 = flow.Process().label('Boxcar smoothing').at((0, 4))
    smooth2 = flow.Process().label('Boxcar smoothing').at((9, 4))
    flow.Arrow().at(smooth1.S).to(box1.N)
    flow.Arrow().at(smooth2.S).to(box2.N)
    
    original2 = flow.Terminal().label('Original Image 2').at((23,8))
    #flow.Arrow().length(3)
   # y = flow.Process().label('calculate standard deviation')
    stdcal1 = flow.Process().label('calculate standard deviation').at((20, 4))
    stdcal2 = flow.Process().label('calculate standard deviation').at((27, 4))


    flow.Arrow().at(stdcal1.S).to(std1.N)
    flow.Arrow().at(stdcal2.S).to(std2.N)
    flow.ArcN(arrow='->').at(original2.S).to(stdcal2.N)
    flow.ArcN(arrow='->').at(original1.S).to(stdcal1.N)
                                                                       
    flow.ArcN(arrow='->').at(original1.S).to(smooth1.N)
    flow.ArcN(arrow='->').at(original2.S).to(smooth2.N)
    flow.Arc2(arrow='->', k = -0.4).at(original2.W).to(subtract2.NNE)
    
    #flow.Arc2(arrow='->').at(original1.S).to(subtract.E)
    #flow.Arrow().at(original1.S).to(subtract.NE)
    flow.Arc2(arrow='->', k = -0.6).at(original1.W).to(subtract.W)
    d.save('NormProd method.pdf')
```

```python
#version two.. (notsplit)

with schemdraw.Drawing() as d:
    d.config(fontsize=13, unit=.5)

    # Column 1
    box1 = flow.Terminal().label('Boxcar 1')

    # Column 2, placed closer (x=5 instead of 17)
    box2 = flow.Terminal().label('Boxcar 2').at((9, 0))
    
    sub = flow.Process().label('Subtract boxcar from original image').at((4.5, -3))

    flow.ArcN(arrow='->').at(box1.S).to(sub.N)
    flow.ArcN(arrow='->').at(box2.S).to(sub.N)

    #DoBs
    dob1 = flow.Terminal().label('Difference of Boxcars 1 (DoB 1)').at((1, -7))
    dob2 = flow.Terminal().label('Difference of Boxcars 2 (DoB 2)').at((9, -7))
    flow.ArcN(arrow='->').at(sub.S).to(dob1.N)
    flow.ArcN(arrow='->').at(sub.S).to(dob2.N)
    
    # column 3 (middle) ?
    a = flow.Process().label('Multiply DoBs together').at((4.5, -10))
    flow.Arrow().length(1.5)
    normprod = flow.Terminal().label('NormProd')
    flow.Arrow().length(1.5)
    mean = flow.Process().label('mean (11x11)')
    flow.Arrow().length(1.5)
    summednormprod = flow.Terminal().label('Summed_Normprod')

    flow.ArcN(arrow='->').at(dob1.S).to(a.N)
    flow.ArcN(arrow='->').at(dob2.S).to(a.N)


    #collumn 4
    std1 = flow.Terminal().label('Stdev 1').at((17, 0))
    std2 = flow.Terminal().label('Stdev 2').at((23, 0))
    mean2 = flow.Process().label('calculate the mean of the two stdevs').at((20, -3.5))
    flow.Arrow().length(1.5)
    stdmean = flow.Terminal().label('Stdmean')
    flow.Arrow().length(1.5)
    flow.Process().label('square Stdmean')
    flow.Arrow().length(1.5)
    flow.Terminal().label('Variance')
    flow.Arrow().length(1.5)
    boxcar2 = flow.Process().label('Boxcar smoothing (11x11)')
    flow.Arrow().length(1.5)
    smovar = flow.Terminal().label('Smoothed_Variance')

    flow.ArcN(arrow='->').at(std1.S).to(mean2.N)
    flow.ArcN(arrow='->').at(std2.S).to(mean2.N)

    #final column
    x = flow.Process().label('Summed_NormProd/Smoothed_Variance').at((12, -23))
    flow.Arrow().length(1.5)
    flow.Terminal().label('NormProd_SmoVar')

    flow.Arc2(arrow='->', k = -0.5).at(summednormprod.S).to(x.W)
    flow.Arc2(arrow='->').at(smovar.S).to(x.E)

    # top columns
    original1 = flow.Terminal().label('Original Image 1').at((4.5,8))
    flow.Arrow().length(3)
    boxcar = flow.Process().label('Boxcar smoothing (11x11)')
    
    
    original2 = flow.Terminal().label('Original Image 2').at((20,8))
    flow.Arrow().length(3)
    y = flow.Process().label('calculate standard deviation')


    flow.ArcN(arrow='->').at(y.S).to(std2.N)
    flow.ArcN(arrow='->').at(y.S).to(std1.N)
    flow.ArcN(arrow='->').at(original1.S).to(y.NNW)
    flow.ArcN(arrow='->').at(original2.S).to(boxcar.NNE)
    flow.ArcN(arrow='->').at(boxcar.S).to(box1.N)
    flow.ArcN(arrow='->').at(boxcar.S).to(box2.N)
    flow.Arc2(arrow='->', k = -0.9).at(original1.W).to(sub.W)
    flow.ArcZ(arrow='->').at(original2.W).to(sub.E)

   # d.save('miz_flowchart_2col.png', dpi=300)
   # d.save('miz_flowchart_2col.eps')
    d.save('NormProd method.pdf')
    
```
