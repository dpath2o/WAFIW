```python
from thredds_crawler.crawl import Crawl
import os
import requests
import pdb
```

```python
# the XML catalog URL (THREDDS provides a machine-readable XML)
catalog_url = "https://tds-odatis.aviso.altimetry.fr/thredds/catalog/dataset-l3-swot-karin-nadir-validated/l3_lr_ssh/v2_0_1/Unsmoothed/cycle_021/catalog.html"

c = Crawl(catalog_url)
datasets = c.datasets  # list of dataset objects

# download directory
os.makedirs("/g/data/gv90/ae8456/SWOT_SEAICE/DATA/SWOT_L3_Unsmoothed_Full_test", exist_ok=True)
```

```python
datasets
```

```python
for ds in datasets:
    pdb.set_trace()
    # get HTTPServer access URL for the .nc file
    for service in ds.services:
        if service["service"] == "HTTPServer":
            url = service["url"]
            local_name = os.path.join("cycle_021_nc", os.path.basename(url))
            print(f"Downloading {url}")
            pdb.set_trace()
            resp = requests.get(url, stream=True)
            with open(local_name, "wb") as f:
                for chunk in resp.iter_content(chunk_size=8192):
                    f.write(chunk)
            break
```
